###**Nested For**

```python
lst = [[2,3,4] , [3,4,5]]
for i in lst :
  print(i)
```

```python
lst = [[2,3,4] , [3,4,5]]
for i in lst :
  print(i)
  for j in i:
    print(j)
  print("########")
```

```python
lst = [[1,2,3],[4,5,6],[7,8,9]]
S=[]
M = 0
for i in lst :
  s = 0
  for j in i :
    s = s + j
  S.append([s])
  M = M + s

print(S)
print(M)
```

```python
# x = 1
# while x <= 10 :  infinite   True
#   print(x)
```

```python
x = 1
while x <= 10 :
  print(x)
  x += 1
```

```python
while True :
  x =  input("Enter char: ")
  if x.isalpha() :
    print(x)
  else:
    print("Wrong char")
    break
```

##**Try Except**

```python
try:
  age = int(input("Please enter your age: "))

except:
  print("Please enter your age in number")

print("s2")
```

```python
try:
  print(10/0)
except ZeroDivisionError:
  print("ZeroDivisionError")
```

```python
try:
  age = int(input("Please enter your age: "))

except ValueError:
  print("Please enter your age in number")

print("s2")
```

```python
try:
  print(10/0)
except Exception as e:
  print("Error", e)
```

```python
try:
  d = {'key':'value'}
  print(d['key2'])
except KeyError:
  print("KeyError")
```
