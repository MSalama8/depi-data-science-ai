```python
def even_odd_sum(lst):
  even = []
  odd = []
  s_even = 0
  s_odd = 0
  for i in lst:
    if i % 2 == 0:
      even.append(i)
      s_even += i
    else:
      odd.append(i)
      s_odd += i
  return even , s_even , odd , s_odd
```

```python
l = [1,2,3,2,3,4,3,4,5,6]
even , s_even , odd , s_odd = even_odd_sum(l)
```

```python
print("Even" , even , "Sum" , s_even)
print("Odd" , odd , "Sum" , s_odd)
```

```python
def get_sum(seq):
  s = 0
  for i in seq:
    s += i
  return s
```

```python
def even_odd_sum(lst):
  even = []
  odd = []
  for i in lst:
    if i % 2 == 0 :
      even.append(i)
    else:
      odd.append(i)
  s_even = get_sum(even)
  s_odd = get_sum(odd)
  return even , s_even , odd , s_odd
```

```python
l = [1,2,3,2,3,4,3,4,5,6]
even , s_even , odd , s_odd = even_odd_sum(l)
```

```python
print("Even" , even , "Sum" , s_even)
print("Odd" , odd , "Sum" , s_odd)
```

```python
def odd_even_lst(lst):
  even = []
  odd = []
  for i in lst:
    if i % 2 == 0 :
      even.append(i)
    else:
      odd.append(i)
  return even , odd
```

```python
def even_odd_sum(lst):
  even , odd = odd_even_lst(lst)
  s_even = get_sum(even)
  s_odd = get_sum(odd)
  return even , s_even , odd , s_odd
```

```python
even , s_even , odd , s_odd = even_odd_sum(l)
```

```python
print("Even" , even , "Sum" , s_even)
print("Odd" , odd , "Sum" , s_odd)
```
