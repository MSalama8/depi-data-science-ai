```python
books = soup.find_all('article' , class_='product_pod')
```

```python
soup.find_all('article' , class_='product_pod')[0].find('p' , class_='instock availability').get_text(strip=True)
```

```python
def extract_book_info(listofbooks):
    book_name = []
    price = []
    availability = []
    for book in listofbooks:
        book_name.append(book.find('h3').get_text())
        price.append(book.find('p' , class_='price_color').get_text())
        availability.append(book.find('p' , class_='instock availability').get_text(strip=True))
    return book_name , price , availability
```

```python
book_name , price , availability = extract_book_info(books)
```

```python
pd.DataFrame({'book_name':book_name , 'price':price , 'availability':availability})
```
